## Exemplu simplu de input si procesare

```python
# functie pentru curatare text
def clean_text(text):
    text = text.lower()
    return text

# functie pentru extragere skill-uri
def extract_skills(text, skills_list):
    found_skills = []

    for skill in skills_list:
        if skill in text:
            found_skills.append(skill)

    return found_skills
```

```python
# exemplu simplu de input

cv_text = """
Python, SQL, Machine Learning, 2 years experience
"""

job_description = """
Looking for Python developer with SQL skills
"""

# lista de skill-uri
skills_list = ["python", "sql", "machine learning", "java"]

# procesare
cv_clean = clean_text(cv_text)
job_clean = clean_text(job_description)

print("CV:", cv_clean)
print("Job:", job_clean)

# extragere skill-uri
cv_skills = extract_skills(cv_clean, skills_list)
job_skills = extract_skills(job_clean, skills_list)

print("Skills CV:", cv_skills)
print("Skills Job:", job_skills)
```

```python
# calcul scor simplu
def compute_match_score(cv_skills, job_skills):
    if len(job_skills) == 0:
        return 0

    matched = set(cv_skills) & set(job_skills)
    score = len(matched) / len(job_skills) * 100

    return round(score, 2), list(matched)

score, matched = compute_match_score(cv_skills, job_skills)

missing = list(set(job_skills) - set(cv_skills))

print("Match Score:", score, "%")
print("Matched Skills:", matched)
print("Missing Skills:", missing)
```

```python
print("Lungime CV:", len(cv_clean))
print("Lungime Job:", len(job_clean))
print("Numar skill-uri CV:", len(cv_skills))
print("Numar skill-uri Job:", len(job_skills))
```

## Observatii

- Textul este procesat prin functii dedicate, ceea ce permite reutilizarea codului.
- Skill-urile sunt extrase folosind o lista predefinita si keyword matching.
- Sistemul poate identifica skill-uri comune intre CV si job description.
- Se calculeaza un scor de potrivire bazat pe procentul de skill-uri regasite.
- Sunt identificate skill-urile lipsa, utile pentru recrutori.

Limitari:
- metoda functioneaza doar pentru potriviri exacte de cuvinte
- nu detecteaza sinonime sau contexte
