## Exemplu simplu de input si procesare

```python
# exemplu simplu de input

cv_text = """
Python, SQL, Machine Learning, 2 years experience
"""

job_description = """
Looking for Python developer with SQL skills
"""

# transformare simpla
cv_text = cv_text.lower()
job_description = job_description.lower()

print("CV:", cv_text)
print("Job:", job_description)

# extragere simpla de skill-uri
skills_list = ["python", "sql", "machine learning", "java"]

found_skills = []

for skill in skills_list:
    if skill in cv_text:
        found_skills.append(skill)

print("Skills gasite:", found_skills)

# analiza simpla
print("Lungime CV:", len(cv_text))
print("Lungime job description:", len(job_description))
print("Numar skill-uri gasite:", len(found_skills))

## Observatii

 #Textul este procesat prin transformare in litere mici.
 #Skill-urile sunt extrase folosind o lista predefinita.
 #Metoda este simpla si functioneaza doar pentru potriviri exacte.
 #In etapele urmatoare se vor folosi metode mai avansate de NLP pentru rezultate mai precise.
```
